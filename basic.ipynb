```python
# Examples of basic data types and variables in Python

# Integer
age = 25
print(f"Age: {age}, Type: {type(age)}")

# Float
height = 5.9
print(f"Height: {height}, Type: {type(height)}")

# String
name = "Alice"
print(f"Name: {name}, Type: {type(name)}")

# Boolean
is_student = True
print(f"Is student: {is_student}, Type: {type(is_student)}")

# List
fruits = ["apple", "banana", "cherry"]
print(f"Fruits: {fruits}, Type: {type(fruits)}")

# Tuple
coordinates = (10, 20)
print(f"Coordinates: {coordinates}, Type: {type(coordinates)}")

# Dictionary
person = {"name": "Bob", "age": 30}
print(f"Person: {person}, Type: {type(person)}")

# Set
unique_numbers = {1, 2, 3, 3}
print(f"Unique numbers: {unique_numbers}, Type: {type(unique_numbers)}")
```

```python
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
# Example of control flow using if statements with existing variables

if age > 20:
    print(f"{name} is older than 20.")
else:
    print(f"{name} is 20 or younger.")

if is_student:
    print("This person is a student.")
else:
    print("This person is not a student.")

if len(fruits) > 2:
    print("There are more than 2 fruits in the list.")
else:
    print("There are 2 or fewer fruits.")

if person['age'] >= 30:
    print(f"{person['name']} is 30 or older.")
else:
    print(f"{person['name']} is younger than 30.")
```

```python
# control flow examples using existing notebook variables

# Age category
if age < 13:
    print(f"{name} is a child.")
elif age < 20:
    print(f"{name} is a teenager.")
elif age < 65:
    print(f"{name} is an adult.")
else:
    print(f"{name} is a senior.")

# Height classification
if height >= 6.0:
    print("Height: tall")
elif height >= 5.5:
    print("Height: average")
else:
    print("Height: short")

# Student status
if is_student:
    print(f"{name} is a student.")
else:
    print(f"{name} is not a student.")

# Fruits count
if len(fruits) == 0:
    print("No fruits.")
elif len(fruits) <= 2:
    print("One or two fruits.")
else:
    print(f"Multiple fruits: {fruits}")

# Compare ages with person dict
if person['age'] > age:
    print(f"{person['name']} is older than {name}.")
elif person['age'] < age:
    print(f"{person['name']} is younger than {name}.")
else:
    print(f"{person['name']} and {name} are the same age.")

# Set membership
if 3 in unique_numbers:
    print("3 is in unique_numbers.")
else:
    print("3 is not in unique_numbers.")
```

```python
# Elif chain examples using existing notebook variables

# 1. Grade from a score (uses variable i)
score = i
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'
print(f"Score {score} -> grade {grade}")

# 2. Fruit count classification (uses fruits)
n_fruits = len(fruits)
if n_fruits == 0:
    size = "no fruits"
elif n_fruits <= 2:
    size = "a couple of fruits"
else:
    size = "many fruits"
print(f"{n_fruits} fruits -> {size}")

# 3. Combined conditions comparing person and existing age/is_student
if person['age'] > age and is_student:
    status = f"{person['name']} is older than {name} and is a student."
elif person['age'] > age:
    status = f"{person['name']} is older than {name} and not a student."
elif person['age'] == age:
    status = f"{person['name']} and {name} are the same age."
else:
    status = f"{person['name']} is younger than {name}."
print(status)

# 4. Coordinate position check (uses coordinates)
x, y = coordinates
if x == 0 and y == 0:
    region = "origin"
elif x == 0:
    region = "on the Y axis"
elif y == 0:
    region = "on the X axis"
elif x < 0 or y < 0:
    region = "has a negative coordinate"
else:
    region = "both coordinates positive"
print(f"Coordinates {coordinates} -> {region}")
```

```python
for j, fruit in enumerate(fruits, start=1):
    print(f"Fruit {j}: {fruit}")
```

```python
# emulate a do-while loop: run body at least once, then repeat while condition holds
while True:
    if 0 <= idx < n_fruits:
        print(f"Fruit {idx+1}: {fruits[idx]}")
    else:
        print(f"No fruit at index {idx}")
    idx += 1
    if not (idx < n_fruits):
        break
```
